## 🏋️ Final exercise: your first kNN, by hand

A **new patient** comes in with `radius_mean = 14.0` and `texture_mean = 20.0`. Is the tumour benign or malignant? Use everything from today:

1. `X`: a numpy array of `radius_mean` and `texture_mean` for **all** tumours (notebook 04, section 5). `y`: the `diagnosis` values as a numpy array.
2. `distances`: the distance from the new patient to **every** tumour, with no loop (notebook 03, section 5).
3. `nearest_5`: the indices of the 5 nearest tumours (`np.argsort`).
4. `prediction`: the most common label among those 5 neighbours. Hint: `pd.Series(y[nearest_5]).value_counts().index[0]`

Then run the plot cell after the check to see your neighbours.
