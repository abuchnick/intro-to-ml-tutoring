## 1. A program is a list of instructions

Python runs instructions **one at a time, top to bottom**. In a notebook, each cell runs when you press **Shift+Enter**.

⚠️ Python remembers what earlier cells did **in the order you ran them, not the order they appear on the page**. If your notebook starts acting strangely, use **Restart → Run All**. That runs everything again from a clean start, which is exactly how anyone else running your notebook will see it.

```python
# ✍️ Print a message
print("Hello, ML!")
```

```python
# ✍️ Python as a calculator. A cell shows the value of its last line.
2 + 3 * 4
```

## 2. Variables are names attached to values

`=` does **not** mean "equals" in the maths sense. It means *"compute the right-hand side, then put the name on the left onto that value."*

```python
# ✍️ Patient #1047 from lecture 1
age = 52
bmi = 31.4
name = "Patient 1047"
print(name, age, bmi)
```

```python
# ✍️ Reassigning: the name moves to a new value
age = age + 1
age
```

```python
# 🤔 Predict: what will this print? Write your guess here: ...
x = 10
y = x
x = 20
print(y)
```

`y = x` gave `y` the **value** that `x` had at that moment (10). Moving `x` to a new value afterwards doesn't change `y`.

## 3. Every value has a type

A value's type decides what you can do with it. You can multiply two numbers, but you can't multiply two pieces of text.

```python
# ✍️ Ask Python for the type of each value
type(age), type(bmi), type(name), type(True)
```

```python
# 🤔 Predict: what happens here?
"5" + 5
```

### How to read an error

**Start at the last line.** It tells you what kind of error happened (`TypeError`) and why. The arrow above it points to the line that failed. Errors are not a disaster, they are Python telling you what went wrong. Learn to read them before you paste them into an AI.

```python
# ✍️ Fix it by converting one value so both have the same type
print(int("5") + 5)
print("5" + str(5))
```

```python
# ✍️ f-strings: put values inside text. `:.1f` means "show 1 decimal place".
print(f"{name} is {age} years old, BMI {bmi:.1f}")
```

## 4. Lists: an ordered collection of values

```python
# ✍️ Fasting glucose readings for four patients
glucose = [118, 95, 142, 101]
glucose[0], glucose[-1], len(glucose)
```

```python
# ✍️ Slicing: [start:stop] includes start but NOT stop
glucose[1:3]
```

```python
# ✍️ Lists can change: add a value and replace a value
glucose.append(130)
glucose[0] = 120
glucose
```

```python
# 🤔 Predict: what does this print?
letters = ["a", "b", "c", "d"]
print(letters[1], letters[-1], letters[:2])
```

```python
# 🤔 Predict: what is `a` at the end?
a = [1, 2, 3]
b = a
b.append(4)
print(a)
```

Surprised? `b = a` did **not** copy the list. It added a second name for the **same** list, so changing it through `b` also changes what `a` shows. If you want an independent copy, use `b = a.copy()`. You'll run into this again with numpy and pandas.

## 5. Dictionaries: look values up by name

```python
# ✍️ One patient as a dictionary: key → value
patient = {"age": 52, "bmi": 31.4, "glucose": 118}
patient["bmi"]
```

```python
# ✍️ Add a new key, then list all the keys
patient["family_history"] = True
patient.keys()
```
