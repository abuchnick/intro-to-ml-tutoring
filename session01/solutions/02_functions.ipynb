# 02 · Decisions, loops and functions

So far, Python has run each line exactly once, top to bottom. This notebook covers the three tools that change that:

- **`if`**: run some code *only when* a condition is true
- **`for`**: run some code *once for each* item in a collection
- **functions**: give a piece of logic a *name*, so you can reuse it and test it

## 1. `if`: making decisions

Remember lecture 1's hand-written rule for diabetes risk? Let's write it as code.

⚠️ **Indentation is part of the code.** The indented lines belong to the `if`, and when the indentation ends, so does the block.

````python
# ✍️ Lecture 1's "Route 1" rule
glucose = 118
bmi = 31.4

if glucose >= 100 and bmi >= 30:
    risk_level = "high"
else:
    risk_level = "low"

print(risk_level)
````

````python
# ✍️ Comparisons produce True or False
glucose >= 100, bmi < 25, glucose == 118, glucose != 118
````

## 2. `for`: doing something once for each item

````python
# ✍️ Loop over a list. `g` refers to each value in turn.
for g in [118, 95, 142, 101]:
    print(g, g >= 100)
````

````python
# ✍️ The "accumulator" pattern: start at 0, add to it each time around the loop
values = [118, 95, 142, 101]
total = 0
for v in values:
    total = total + v
mean_glucose = total / len(values)
mean_glucose
````

````python
# 🤔 Predict: go through the loop by hand, one value at a time. What gets printed?
count = 0
for v in [118, 95, 142, 101]:
    if v >= 100:
        count += 1      # same as count = count + 1
print(count)
````

````python
# ✍️ range(n) gives the numbers 0, 1, ..., n-1. enumerate gives each index and value together.
for i in range(3):
    print("round", i)

for i, g in enumerate(values):
    print(f"patient {i}: glucose {g}")
````

## 3. Functions: name a piece of logic

A function takes **inputs** (its *parameters*), runs some code, and **`return`s** an output.

Why bother writing functions?
1. **Reuse**: you write the logic once and use it many times.
2. **Readability**: `risk(118, 31.4)` says what it does.
3. **Testing**: you can check a small piece of code on its own.

````python
# ✍️ The risk rule as a function
def risk(glucose, bmi):
    """Return 'high' or 'low' diabetes risk, using lecture 1's hand-written rule."""
    if glucose >= 100 and bmi >= 30:
        return "high"
    return "low"

risk(118, 31.4), risk(90, 25)
````

````python
# 🤔 Predict: what do x and y end up holding? (Careful, this one is tricky.)
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

x = add_print(2, 3)
y = add_return(2, 3)
print("x =", x, "| y =", y)
````

`print` only **shows** a value on the screen. `return` **hands the value back** to the code that called the function. A function without a `return` gives back `None`. If you need to use the result later in your code, you need `return`.

````python
# 🤔 Predict: does the last line work?
def make_variable():
    inside = 42
    return inside

make_variable()
print(inside)
````

Variables created inside a function only exist **while the function is running** (this is called *scope*). The only way to get a value out of a function is to `return` it.

````python
# ✍️ Testing with assert: it does nothing if the condition is True and raises an error if it's False
assert risk(118, 31.4) == "high"
assert risk(90, 25) == "low"
assert risk(100, 30) == "high"   # the edge case: are the boundaries >= or >?
print("all tests passed ✅")
````

## 🏋️ Exercise 1: `mean`

Write a function `mean(values)` that returns the average of a list of numbers. Use a `for` loop, **not** the built-in `sum()`.

````python
def mean(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)
````

````python
# --- Check (do not edit) ---
assert mean([1, 2, 3]) == 2, "mean([1, 2, 3]) should be 2. Did you remember to `return`?"
assert mean([118, 95, 142, 101]) == 114
print("✅ Exercise 1 correct!")
````

## 🏋️ Exercise 2: `euclidean_distance` (the heart of kNN!)

Remember kNN from the recitation? It classifies a new patient by finding the training patients that are **closest** to them. "Closest" is measured with the Euclidean distance:

$$d(a, b) = \sqrt{(a_1-b_1)^2 + (a_2-b_2)^2 + \dots + (a_n-b_n)^2}$$

Write `euclidean_distance(a, b)` for two lists of the same length. Hints: `x ** 2` squares a number and `x ** 0.5` takes its square root. To walk through both lists at the same time, use `for i in range(len(a)):`.

````python
def euclidean_distance(a, b):
    total = 0
    for i in range(len(a)):
        total += (a[i] - b[i]) ** 2
    return total ** 0.5
````

````python
# --- Check (do not edit) ---
assert euclidean_distance([0, 0], [3, 4]) == 5, "the distance from (0,0) to (3,4) should be 5"
assert euclidean_distance([1, 2, 3], [1, 2, 3]) == 0, "a point's distance to itself should be 0"
assert round(euclidean_distance([17.99, 10.38], [20.57, 17.77]), 2) == 7.83
print("✅ Exercise 2 correct! You just wrote the core of kNN.")
````
